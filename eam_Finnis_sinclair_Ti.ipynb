```python
%%writefile Ti_eam.in

clear

units metal
dimension 3
boundary p p p
atom_style atomic

lattice hcp 2.95
region box block 0 10 0 10 0 10
create_box 1 box
create_atoms 1 box

pair_style eam/fs
pair_coeff * * Ti1.eam.fs Ti

neighbor 2.0 bin
neigh_modify every 1 delay 0 check yes

thermo 100
thermo_style custom step temp pe ke etotal press vol lx ly lz
```
